```python
players['date_of_birth']=pd.to_datetime(players['date_of_birth'],errors='coerce')
```

```python
birth_year = players["date_of_birth"].dt.year
birth_month = players["date_of_birth"].dt.month
birth_day = players["date_of_birth"].dt.day
```

```python
current_year=2026
current_month=11
current_date=2
```

```python
age=2026-birth_year
```

```python
age
```

```python
condition = ((birth_day>current_date)&(birth_month==current_month))|(birth_month>current_month)
```

```python
age.loc[condition==True]=age.loc[condition==True]-1
```

```python
age
```

```python
players['age']=age
```

```python
# youngest player
players[players['age']==players['age'].min()][['name','age']]
```

```python
players['age_group']=pd.cut(
    players['age'],
    bins=[0,17,21,25,30,100],
    labels=['Young Prospects','Developing','Prime Potential','Established','Veteran'],
)
```

```python
import seaborn as sns
```

```python
ax=sns.boxplot(players,x='age_group',y='market_value_in_eur')
ax.set_yscale("log")
```

```python
sc=sns.scatterplot(players,x='age',y='market_value_in_eur')
sc.set_yscale("log")
```

Observation 1: Peak market value (approximately 20–30 years)
 Players between 20 and 30 have the highest market valuations.
 The highest-valued players are concentrated around 22–28.
 This broadly corresponds to the prime years of a footballer's career.

Observation 2: Decline after 30
 After approximately 30, the upper range of market valuations starts declining.
 Players above 35 generally have lower market valuations.
 This could be due to declining physical performance, shorter remaining career length and reduced resale potential.

Observation 3: Variation at the same age
 If we look btw ages 24-28 we see that some players have exceptionally high market value while some players have very less market value.
 # This tells us that age alone cannot determine a player's market value
