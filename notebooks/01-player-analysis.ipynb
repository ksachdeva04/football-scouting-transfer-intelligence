```python
players['date_of_birth']=pd.to_datetime(players['date_of_birth'],errors='coerce')
```

```python
players['date_of_birth'].dtype
```

```python
birth_year = players["date_of_birth"].dt.year
birth_month = players["date_of_birth"].dt.month
birth_day = players["date_of_birth"].dt.day
```

```python
current_year=2026
current_month=11
current_date=2
```

```python
age=2026-birth_year
```

```python
age
```

```python
condition = ((birth_day>current_date)&(birth_month==current_month))|(birth_month>current_month)
```

```python
age.loc[condition==True]=age.loc[condition==True]-1
```

```python
age
```

```python
players['age']=age
```

```python
# youngest player
players[players['age']==players['age'].min()][['name','age']]
```

```python
# oldest player
players[players['age']==players['age'].max()][['name','age','position','date_of_birth']]
```

```python
players['age'].isnull().sum()
```

```python
players['age_group']=pd.cut(
    players['age'],
    bins=[0,17,21,25,30,100],
    labels=['Young Prospects','Developing','Prime Potential','Established','Veteran'],
)
```

```python
players['age_group'].value_counts().sort_values(ascending=False)
```

```python
players.groupby(by='age_group')['market_value_in_eur'].mean()
```
